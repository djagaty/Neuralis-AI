# Assignment 8: Pivot Tables

```python
import numpy as np
import pandas as pd


def pivot_sum(df):
    return df.pivot_table(index="Date", columns="Category", values="Value", aggfunc="sum", fill_value=0)


def pivot_mean_revenue(df):
    return df.pivot_table(index="Year", columns="Quarter", values="Revenue", aggfunc="mean")
```

```python
# sample answer
df1 = pd.DataFrame({
    "Date": ["2021-01-01", "2021-01-01", "2021-01-02", "2021-01-02"],
    "Category": ["A", "B", "A", "B"],
    "Value": [10, 20, 30, 40],
})
print(pivot_sum(df1))

df2 = pd.DataFrame({
    "Year": [2020, 2020, 2021, 2021, 2021],
    "Quarter": ["Q1", "Q2", "Q1", "Q2", "Q2"],
    "Revenue": [100, 150, 200, 250, 300],
})
print(pivot_mean_revenue(df2))
```

```python
# edge and negative case checks
df = pd.DataFrame({
    "Date": ["2021-01-01", "2021-01-01", "2021-01-02"],
    "Category": ["A", "B", "A"],
    "Value": [10, 20, 30],
})
pivot = pivot_sum(df)
assert pivot.loc["2021-01-01", "A"] == 10
assert pivot.loc["2021-01-01", "B"] == 20
assert pivot.loc["2021-01-02", "A"] == 30
assert pivot.loc["2021-01-02", "B"] == 0

single = pd.DataFrame({"Date": ["2021-01-01"], "Category": ["A"], "Value": [5]})
assert pivot_sum(single).loc["2021-01-01", "A"] == 5

try:
    pivot_sum(pd.DataFrame({"Date": ["2021-01-01"], "Category": ["A"]}))
    assert False, "expected KeyError for a missing Value column"
except KeyError:
    pass

df2 = pd.DataFrame({
    "Year": [2020, 2020, 2021, 2021],
    "Quarter": ["Q1", "Q1", "Q1", "Q2"],
    "Revenue": [100, 200, 300, 400],
})
pivot2 = pivot_mean_revenue(df2)
assert pivot2.loc[2020, "Q1"] == 150
assert pivot2.loc[2021, "Q1"] == 300
assert pd.isna(pivot2.loc[2020, "Q2"])

try:
    pivot_mean_revenue(pd.DataFrame({"Year": [2020], "Quarter": ["Q1"]}))
    assert False, "expected KeyError for a missing Revenue column"
except KeyError:
    pass

print("all tests passed")
```
