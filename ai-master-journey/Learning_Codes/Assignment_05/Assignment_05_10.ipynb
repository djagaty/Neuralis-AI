# Assignment 10: Working with Text Data

```python
import numpy as np
import pandas as pd


def to_upper(series):
    return series.str.upper()


def first_three_chars(series):
    return series.str[:3]
```

```python
# sample answer
words = pd.Series(["apple", "Banana", "cherry", "kiwi", "Mango"])
upper = to_upper(words)
first_three = first_three_chars(words)
print(upper)
print(first_three)
```

```python
# edge and negative case checks
series = pd.Series(["apple", "Banana", "CHERRY", "kiwi", "Mango"])
upper = to_upper(series)
assert list(upper) == ["APPLE", "BANANA", "CHERRY", "KIWI", "MANGO"]

with_empty = pd.Series(["", "abc"])
assert list(to_upper(with_empty)) == ["", "ABC"]

with_nan = pd.Series(["abc", np.nan])
result = to_upper(with_nan)
assert result.iloc[0] == "ABC"
assert pd.isna(result.iloc[1])

series2 = pd.Series(["hello", "hi", "world", "ab", ""])
result2 = first_three_chars(series2)
assert list(result2) == ["hel", "hi", "wor", "ab", ""]

exact = pd.Series(["abc"])
assert list(first_three_chars(exact)) == ["abc"]

with_nan2 = pd.Series(["hello", np.nan])
result3 = first_three_chars(with_nan2)
assert result3.iloc[0] == "hel"
assert pd.isna(result3.iloc[1])

print("all tests passed")
```
