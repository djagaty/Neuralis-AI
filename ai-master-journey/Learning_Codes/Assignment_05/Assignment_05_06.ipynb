# Assignment 6: Time Series Analysis

```python
import numpy as np
import pandas as pd


def make_time_series(start, periods, freq, rng, low=1, high=100):
    idx = pd.date_range(start=start, periods=periods, freq=freq)
    values = rng.integers(low, high + 1, size=periods)
    return pd.Series(values, index=idx)


def monthly_mean(series):
    return series.resample("ME").mean()


def rolling_week_mean(series):
    return series.rolling(window=7).mean()
```

```python
# sample answer
rng = np.random.default_rng(42)
series1 = make_time_series("2021-01-01", 90, "D", rng)
monthly = monthly_mean(series1)
print(monthly)

full_year_index = pd.date_range("2021-01-01", "2021-12-31", freq="D")
series2 = pd.Series(rng.integers(1, 100, size=len(full_year_index)), index=full_year_index)
rolled = rolling_week_mean(series2)
print(rolled.head(10))
```

```python
# edge and negative case checks
idx = pd.date_range("2021-01-01", periods=10, freq="D")
series = pd.Series(range(1, 11), index=idx)
monthly = monthly_mean(series)
assert len(monthly) == 1
assert monthly.iloc[0] == series.mean()

idx2 = pd.date_range("2021-01-25", periods=10, freq="D")
series2 = pd.Series(range(1, 11), index=idx2)
monthly2 = monthly_mean(series2)
assert len(monthly2) == 2

try:
    monthly_mean(pd.Series([1, 2, 3]))
    assert False, "expected TypeError without a datetime index"
except TypeError:
    pass

full_year_index = pd.date_range("2021-01-01", "2021-12-31", freq="D")
rng2 = np.random.default_rng(5)
series3 = pd.Series(rng2.integers(1, 100, size=len(full_year_index)), index=full_year_index)
rolled = rolling_week_mean(series3)
assert rolled.iloc[:6].isna().all()
assert not pd.isna(rolled.iloc[6])
assert abs(rolled.iloc[6] - series3.iloc[0:7].mean()) < 1e-9

short_series = pd.Series([1, 2, 3], index=pd.date_range("2021-01-01", periods=3, freq="D"))
assert rolling_week_mean(short_series).isna().all()

try:
    series3.rolling(window=-1)
    assert False, "expected ValueError for a negative window"
except ValueError:
    pass

print("all tests passed")
```
