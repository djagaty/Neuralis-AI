# Assignment 3: Data Cleaning

```python
import numpy as np
import pandas as pd


def fill_na_with_mean(df):
    return df.fillna(df.mean(numeric_only=True))


def drop_rows_with_na(df):
    return df.dropna()
```

```python
# sample answer
rng = np.random.default_rng(42)
df1 = pd.DataFrame(rng.integers(1, 50, size=(5, 3)).astype(float), columns=["a", "b", "c"])
na_mask = rng.random(df1.shape) < 0.2
df1[na_mask] = np.nan
filled = fill_na_with_mean(df1)
print(df1)
print(filled)

df2 = pd.DataFrame(rng.integers(1, 50, size=(6, 4)).astype(float), columns=["a", "b", "c", "d"])
na_mask2 = rng.random(df2.shape) < 0.2
df2[na_mask2] = np.nan
dropped = drop_rows_with_na(df2)
print(df2)
print(dropped)
```

```python
# edge and negative case checks
df = pd.DataFrame({"a": [1.0, np.nan, 3.0], "b": [4.0, 5.0, np.nan]})
filled = fill_na_with_mean(df)
assert filled.isna().sum().sum() == 0
assert filled.loc[1, "a"] == df["a"].mean()
assert filled.loc[2, "b"] == df["b"].mean()

all_nan = pd.DataFrame({"a": [np.nan, np.nan]})
still_nan = fill_na_with_mean(all_nan)
assert still_nan["a"].isna().all()

no_na = pd.DataFrame({"a": [1.0, 2.0]})
assert fill_na_with_mean(no_na).equals(no_na)

df2 = pd.DataFrame({
    "a": [1, 2, np.nan],
    "b": [4, np.nan, 6],
    "c": [7, 8, 9],
    "d": [1, 2, 3],
})
dropped = drop_rows_with_na(df2)
assert dropped.shape[0] == 1

no_na_df = pd.DataFrame({"a": [1, 2], "b": [3, 4]})
assert drop_rows_with_na(no_na_df).equals(no_na_df)

all_na_rows = pd.DataFrame({"a": [np.nan, 1], "b": [1, np.nan]})
assert drop_rows_with_na(all_na_rows).shape[0] == 0

print("all tests passed")
```
