# Assignment 1: DataFrame Creation and Indexing

```python
import numpy as np
import pandas as pd


def make_df_first_col_index(rows, cols, low, high, rng):
    names = [f"col{i}" for i in range(cols)]
    data = rng.integers(low, high + 1, size=(rows, cols))
    df = pd.DataFrame(data, columns=names)
    return df.set_index(names[0])


def make_labeled_df(rng, low=1, high=100):
    data = rng.integers(low, high + 1, size=(3, 3))
    return pd.DataFrame(data, columns=["A", "B", "C"], index=["X", "Y", "Z"])


def get_value(df, row, col):
    return df.loc[row, col]
```

```python
# sample answer
rng = np.random.default_rng(42)
df1 = make_df_first_col_index(6, 4, 1, 50, rng)
df2 = make_labeled_df(rng)
value = get_value(df2, "Y", "B")
print(df1)
print(df2)
print("value at Y, B:", value)
```

```python
# edge and negative case checks
small = make_df_first_col_index(1, 2, 1, 10, np.random.default_rng(1))
assert small.shape == (1, 1)
assert small.index.name == "col0"

main_df = make_df_first_col_index(6, 4, 1, 50, np.random.default_rng(2))
assert main_df.shape == (6, 3)
assert main_df.index.name == "col0"

labeled = make_labeled_df(np.random.default_rng(3))
assert get_value(labeled, "Y", "B") == labeled.loc["Y", "B"]

try:
    get_value(labeled, "Q", "B")
    assert False, "expected KeyError for a missing row label"
except KeyError:
    pass

print("all tests passed")
```
