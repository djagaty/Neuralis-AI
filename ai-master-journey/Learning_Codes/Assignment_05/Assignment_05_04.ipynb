# Assignment 4: Data Aggregation

```python
import numpy as np
import pandas as pd


def aggregate_by_category(df):
    return df.groupby("Category")["Value"].agg(["sum", "mean"])


def total_sales_by_category(df):
    return df.groupby("Category")["Sales"].sum()
```

```python
# sample answer
rng = np.random.default_rng(42)
categories = rng.choice(["A", "B", "C"], size=10)
values = rng.integers(1, 100, size=10)
df1 = pd.DataFrame({"Category": categories, "Value": values})
agg = aggregate_by_category(df1)
print(df1)
print(agg)

products = [f"p{i}" for i in range(8)]
cats = rng.choice(["X", "Y"], size=8)
sales = rng.integers(10, 500, size=8)
df2 = pd.DataFrame({"Product": products, "Category": cats, "Sales": sales})
totals = total_sales_by_category(df2)
print(df2)
print(totals)
```

```python
# edge and negative case checks
df = pd.DataFrame({"Category": ["A", "B", "A", "C", "B"], "Value": [10, 20, 30, 40, 50]})
agg = aggregate_by_category(df)
assert agg.loc["A", "sum"] == 40 and agg.loc["A", "mean"] == 20
assert agg.loc["B", "sum"] == 70 and agg.loc["B", "mean"] == 35
assert agg.loc["C", "sum"] == 40 and agg.loc["C", "mean"] == 40

single = pd.DataFrame({"Category": ["A", "A"], "Value": [5, 15]})
agg_single = aggregate_by_category(single)
assert agg_single.shape[0] == 1

try:
    aggregate_by_category(pd.DataFrame({"Value": [1, 2]}))
    assert False, "expected KeyError for a missing Category column"
except KeyError:
    pass

df2 = pd.DataFrame({
    "Product": ["p1", "p2", "p3"],
    "Category": ["X", "X", "Y"],
    "Sales": [100, 200, 300],
})
totals = total_sales_by_category(df2)
assert totals["X"] == 300 and totals["Y"] == 300

empty = pd.DataFrame({"Category": [], "Sales": []})
assert total_sales_by_category(empty).shape[0] == 0

try:
    total_sales_by_category(pd.DataFrame({"Category": ["X"]}))
    assert False, "expected KeyError for a missing Sales column"
except KeyError:
    pass

print("all tests passed")
```
