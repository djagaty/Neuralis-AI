# Assignment 9: Applying Functions

```python
import numpy as np
import pandas as pd


def double_values(df):
    return df.apply(lambda col: col * 2)


def add_sum_column(df):
    result = df.copy()
    result["sum"] = df.apply(lambda row: row.sum(), axis=1)
    return result
```

```python
# sample answer
rng = np.random.default_rng(42)
df1 = pd.DataFrame(rng.integers(1, 20, size=(5, 3)), columns=["a", "b", "c"])
doubled = double_values(df1)
print(df1)
print(doubled)

df2 = pd.DataFrame(rng.integers(1, 20, size=(6, 3)), columns=["x", "y", "z"])
with_sum = add_sum_column(df2)
print(with_sum)
```

```python
# edge and negative case checks
df = pd.DataFrame({"a": [1, 2], "b": [3, 4], "c": [5, 6]})
doubled = double_values(df)
assert doubled.equals(pd.DataFrame({"a": [2, 4], "b": [6, 8], "c": [10, 12]}))

empty = pd.DataFrame({"a": [], "b": []})
assert double_values(empty).shape == (0, 2)

try:
    double_values([1, 2, 3])
    assert False, "expected AttributeError for a non DataFrame input"
except AttributeError:
    pass

df2 = pd.DataFrame({"x": [1, 2, 3], "y": [4, 5, 6]})
result = add_sum_column(df2)
assert list(result["sum"]) == [5, 7, 9]
assert "sum" not in df2.columns

single_col = pd.DataFrame({"x": [1, 2, 3]})
assert list(add_sum_column(single_col)["sum"]) == [1, 2, 3]

try:
    add_sum_column([1, 2, 3])
    assert False, "expected AttributeError for a non DataFrame input"
except AttributeError:
    pass

print("all tests passed")
```
