# Assignment 2: DataFrame Operations

```python
import numpy as np
import pandas as pd


def add_product_column(df):
    cols = df.columns
    result = df.copy()
    result["product"] = df[cols[0]] * df[cols[1]]
    return result


def row_and_column_sums(df):
    return df.sum(axis=1), df.sum(axis=0)
```

```python
# sample answer
rng = np.random.default_rng(42)
df1 = pd.DataFrame(rng.integers(1, 20, size=(5, 3)), columns=["a", "b", "c"])
with_product = add_product_column(df1)
print(with_product)

df2 = pd.DataFrame(rng.integers(1, 20, size=(4, 3)), columns=["x", "y", "z"])
row_sums, col_sums = row_and_column_sums(df2)
print(df2)
print("row sums:\n", row_sums)
print("column sums:\n", col_sums)
```

```python
# edge and negative case checks
df = pd.DataFrame({"a": [1, 2, 3], "b": [4, 5, 6], "c": [7, 8, 9]})
result = add_product_column(df)
assert list(result["product"]) == [4, 10, 18]
assert "product" not in df.columns

two_col = pd.DataFrame({"a": [1, 2], "b": [3, 4]})
assert list(add_product_column(two_col)["product"]) == [3, 8]

try:
    add_product_column(pd.DataFrame({"a": [1, 2]}))
    assert False, "expected IndexError with a single column"
except IndexError:
    pass

df2 = pd.DataFrame({"x": [1, 2], "y": [3, 4], "z": [5, 6]})
row_sums, col_sums = row_and_column_sums(df2)
assert list(row_sums) == [9, 12]
assert col_sums["x"] == 3 and col_sums["y"] == 7 and col_sums["z"] == 11

empty_rows, empty_cols = row_and_column_sums(pd.DataFrame())
assert len(empty_rows) == 0 and len(empty_cols) == 0

print("all tests passed")
```
