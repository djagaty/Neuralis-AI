# Assignment 5: Merging DataFrames

```python
import pandas as pd


def merge_on_common(df1, df2, on):
    return pd.merge(df1, df2, on=on)


def concat_rows_and_columns(df1, df2):
    rows = pd.concat([df1, df2], axis=0, ignore_index=True)
    columns = pd.concat([df1, df2], axis=1)
    return rows, columns
```

```python
# sample answer
left = pd.DataFrame({"id": [1, 2, 3, 4], "name": ["a", "b", "c", "d"]})
right = pd.DataFrame({"id": [2, 3, 4, 5], "score": [90, 80, 70, 60]})
merged = merge_on_common(left, right, "id")
print(merged)

df_a = pd.DataFrame({"x": [1, 2, 3]})
df_b = pd.DataFrame({"y": [4, 5, 6]})
rows, columns = concat_rows_and_columns(df_a, df_b)
print(rows)
print(columns)
```

```python
# edge and negative case checks
left = pd.DataFrame({"id": [1, 2, 3], "name": ["a", "b", "c"]})
right = pd.DataFrame({"id": [2, 3, 4], "score": [90, 80, 70]})
merged = merge_on_common(left, right, "id")
assert list(merged["id"]) == [2, 3]
assert merged.loc[merged["id"] == 2, "score"].iloc[0] == 90

no_overlap = merge_on_common(left, pd.DataFrame({"id": [100], "score": [1]}), "id")
assert no_overlap.shape[0] == 0

try:
    merge_on_common(left, pd.DataFrame({"other": [1, 2]}), "id")
    assert False, "expected KeyError for a missing merge column"
except KeyError:
    pass

df_a = pd.DataFrame({"x": [1, 2]})
df_b = pd.DataFrame({"y": [3, 4]})
rows, columns = concat_rows_and_columns(df_a, df_b)
assert rows.shape == (4, 2)
assert columns.shape == (2, 2)
assert list(columns.columns) == ["x", "y"]

try:
    concat_rows_and_columns(df_a, [1, 2, 3])
    assert False, "expected TypeError when concatenating a non DataFrame"
except TypeError:
    pass

print("all tests passed")
```
